```python
import pandas as pd
import pymysql as py
from sqlalchemy import create_engine
engine = create_engine(
    "mysql+pymysql://root:nitya@localhost/tourisms"
)
query = """
SELECT *
FROM tourism;
"""

df = pd.read_sql(query, engine)
```

```python
df_num=df.select_dtypes(include=['int64', 'float64'])
df_num.drop(['AttractionCountryId', 'AttractionRegionId', 'AttractionContinentId',"CountryId", 'RegionId', 'ContinentId','TransactionId'], axis=1, inplace= True)
```

```python
df_num.columns
```

```python
from kmodes.kprototypes import KPrototypes

kproto = KPrototypes(
    n_clusters=3,
    random_state=42
)

clusters = kproto.fit_predict(
    df_num,
    categorical=[0,1,2,3,8]
)
```

```python
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)

sil = silhouette_score(df_num, clusters)
dbi = davies_bouldin_score(df_num, clusters)
ch = calinski_harabasz_score(df_num, clusters)

print(f"Silhouette Score      : {sil:.4f}")
print(f"Davies-Bouldin Score  : {dbi:.4f}")
print(f"Calinski-Harabasz     : {ch:.4f}")
```

```python
print(f"K-Prototypes Cost : {kproto.cost_:,.2f}")
```
